Tipos de Dados com Pandas

```python
display(cidades.dtypes)

cidades.info()
```

```python
cidades['N Habitantes'] = cidades['N Habitantes'].astype('float64')

cidades.dtypes
```

```python
cidades['data'] = [
    '3-8-2021',
    '3-9-2021',
    '3-10-2021',
    '3-11-2021',
    '3-12-2021'
]

cidades['data'] = pd.to_datetime(
    cidades['data'],
    format='%d-%m-%Y'
)

cidades
```

```python
cidades['Data'] = cidades['Data'].dt.strftime('%d-%m-%Y') # Dia, Mês, Ano
cidades # Novo Formato
```

Adicionando e Removendo Colunas

```python
cidades.insert(2, 'sigla', ['PR', 'SP', 'SC', 'PR', 'SP'])
cidades.insert(2, 'siglaRepetida', cidades['sigla'])
cidades
```

Operações Básicas - Combinando DataFrames

```python
cidadesSalarios = pd.DataFrame(
 [
 ['Paraná', 'Londrina', 1356.00],
 ['São Paulo', 'São Carlos', 1508.00],
 ['Santa Catarina', 'Florianópolis', 1798.00],
 ['Paraná', 'Curitiba', 2293.00],
 ['São Paulo', 'Campinas', 1710.00]
 ], columns=['estado', 'cidade', 'salarioMedio'])
cidadesSalarios
```

```python
cidades_pop = pd.DataFrame(
 [
 ['Londrina', 575377],
 ['São Carlos', 254484],
 ['Florianópolis', 508826],
 ['Curitiba', 1963726],
 ['Campinas', 1223237]
 ], columns=['Cidade', 'Habitantes'])
cidades_pop
```

```python
pd.merge(left=cidadesSalarios, right=cidades_pop, left_on="cidade", right_on="Cidade")
```

```python
pd.merge(left=cidadesSalarios, right=cidades_pop, left_on="cidade", right_on="Cidade", how="outer")
```

```python
cidades_df1 = pd.DataFrame(
 [
 ['Londrina', 575377],
 ['São Carlos', 254484],
 ['Florianópolis', 508826],
 ], columns=['Cidade', 'Habitantes'])
cidades_df1
```

```python
cidades_df2 = pd.DataFrame(
 [
 ['Curitiba', 1963726],
 ['Campinas', 1223237]
 ], columns=['Cidade', 'Habitantes'])
cidades_df2
```

```python
pd.concat([cidades_df1, cidades_df2], ignore_index=True)
```
